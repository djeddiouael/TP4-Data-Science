```python
import pandas as pd
import numpy as np
```

```python
# 1. Création du DataFrame
data = {
    'Ville': ['Alger', 'Annaba', 'Oran', 'Alger', 'Oran', 'Annaba', 'Alger', 'Oran'],
    'Service': ['Cardiologie', 'Neurologie', 'Orthopédie', 'Cardiologie', 
                'Neurologie', 'Orthopédie', 'Cardiologie', 'Neurologie'],
    'Age': [45, 60, 30, 50, 40, 70, 35, 55],
    'JoursHospitalis': [5, 8, 3, 7, 4, 6, 2, 9],
    'CoûtJournalier': [200, 300, 150, 220, 280, 160, 210, 290],
    'Satisfaction': [4.5, 3.8, 4.2, 4.0, 4.1, 3.5, 4.7, 3.9],
    'Réadmission': [0, 1, 0, 0, 1, 1, 0, 1]
}

df = pd.DataFrame(data)
print("DataFrame initial:")
print(df)
```

```python
# 2. Ajout de colonnes
df['CoûtTotal'] = df['JoursHospitalis'] * df['CoûtJournalier']

def categoriser_age(age):
    if age < 30:
        return "Jeune"
    elif 30 <= age <= 50:
        return "Moyen"
    else:
        return "Sénior"

df['GroupeÂge'] = df['Age'].apply(categoriser_age)

print("\nAprès ajout de CoûtTotal et GroupeÂge:")
print(df)

```

```python
# 3. Ajouter, renommer et supprimer des colonnes
df['Bonus'] = df['CoûtJournalier'] * 0.1
df = df.rename(columns={'CoûtJournalier': 'TarifJournalier'})
df = df.drop('Bonus', axis=1)

print("\nAprès renommage et suppression:")
print(df.columns)
```

```python
# 4. Transformation des données
df['CoûtÉlevé'] = df['CoûtTotal'].apply(lambda x: 'Oui' if x > 60000 else 'Non')

def categoriser_satisfaction(score):
    if score >= 8:
        return "Excellent"
    elif 5 <= score < 8:
        return "Moyen"
    else:
        return "Faible"

df['NiveauSatisfaction'] = df['Satisfaction'].apply(categoriser_satisfaction)

print("\nAprès transformation des données:")
print(df[['CoûtTotal', 'CoûtÉlevé', 'Satisfaction', 'NiveauSatisfaction']])
```

```python
# 5. Classement et tri
df['Classement'] = df['CoûtTotal'].rank(ascending=False, method='dense').astype(int)

# Calcul du ScoreRisque pour le tri
df['ScoreRisque'] = (df['JoursHospitalis'] * 0.4) + (df['Age'] * 0.3) + ((100 - df['Satisfaction'] * 10) * 0.3)
df = df.sort_values('ScoreRisque', ascending=False)

print("\nAprès classement et tri par ScoreRisque:")
print(df[['Ville', 'CoûtTotal', 'Classement', 'ScoreRisque']])
```

```python
# 6. Résumé par ville
ville_summary = df.groupby('Ville').agg({
    'CoûtTotal': 'mean',
    'Satisfaction': 'mean',
    'Réadmission': 'mean'  # Pourcentage car 0/1
}).round(2)

ville_summary['TauxRéadmission'] = (ville_summary['Réadmission'] * 100).round(2)
ville_summary = ville_summary.drop('Réadmission', axis=1)
ville_summary = ville_summary.rename(columns={
    'CoûtTotal': 'CoûtTotalMoyen',
    'Satisfaction': 'SatisfactionMoyenne'
})

print("\nRésumé par ville:")
print(ville_summary)
```

```python
# 7. Calcul du score de risque et catégorisation
def categoriser_risque(score):
    if score > 60:
        return "Élevé"
    elif 40 <= score <= 60:
        return "Moyen"
    else:
        return "Faible"

df['CatégorieRisque'] = df['ScoreRisque'].apply(categoriser_risque)

print("\nAvec ScoreRisque et CatégorieRisque:")
print(df[['Ville', 'ScoreRisque', 'CatégorieRisque']])

```

```python
# 8. Ville avec le score de risque moyen le plus élevé
ville_risque_moyen = df.groupby('Ville')['ScoreRisque'].mean()
ville_risque_max = ville_risque_moyen.idxmax()

print(f"\nVille avec le score de risque moyen le plus élevé: {ville_risque_max}")
print(f"Scores de risque moyens par ville:\n{ville_risque_moyen}")
```

```python
# 9. Nettoyage et mapping des données
# Remplacer les valeurs manquantes (s'il y en a)
if df['Satisfaction'].isnull().any():
    df['Satisfaction'] = df['Satisfaction'].fillna(df['Satisfaction'].mean())

# Mapping des villes vers codes de région
mapping_villes = {'Alger': 'DZ-16', 'Annaba': 'DZ-23', 'Oran': 'DZ-31'}
df['CodeRégion'] = df['Ville'].map(mapping_villes)

# Mapping de la réadmission
df['Réadmission'] = df['Réadmission'].map({0: 'Non', 1: 'Oui'})

print("\nAprès nettoyage et mapping:")
print(df[['Ville', 'CodeRégion', 'Réadmission']])


```

```python
# 10. Analyse par tableau croisé
pivot_table = df.pivot_table(
    values=['CoûtTotal', 'Satisfaction'],
    index=['Ville', 'Service'],
    aggfunc='mean'
).round(2)

grouped_summary = df.groupby(['Ville', 'Service']).agg({
    'CoûtTotal': 'mean',
    'Satisfaction': 'mean'
}).round(2)

print("\nTableau croisé dynamique:")
print(pivot_table)
print("\nRésumé groupé:")
print(grouped_summary)
```

```python
# 11. Exportation des résultats
df.to_csv('hospital_risk.csv', index=False)
ville_summary.to_csv('city_summary.csv')

print("\nExportation terminée: hospital_risk.csv et city_summary.csv")
```
