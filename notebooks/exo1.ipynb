```python
import numpy as np

T = np.array([
    [22, 25, 20],
    [24, 27, 21],
    [19, 20, 22],
    [25, 29, 28],
    [26, 30, 27],
    [21, 21, 23],
    [20, 26, 25]
])

print(T)
```

```python
max_per_city = np.max(T, axis=0)  # [26, 30, 28]
print(max_per_city)
```

```python
max_per_day = np.max(T, axis=1)  # [25, 27, 22, 29, 30, 23, 26]
print(max_per_day)
```

```python
overall_max = np.max(T)  # 30
print(overall_max)
```

```python
overall_max_idx = np.unravel_index(np.argmax(T), T.shape)
print(overall_max_idx)
# (jour, ville) = (4, 1) → Jour 5, ville Annaba
```

```python
max_alger_oran = np.maximum(T[:, 0], T[:, 2])  # [22, 24, 22, 28, 27, 23, 25]
print(max_alger_oran)
```

```python
mean_per_city = np.mean(T, axis=0)  # [22.43, 25.43, 23.71]
T_with_mean = np.vstack([T, mean_per_city])
print(mean_per_city)
print(T_with_mean)
```

```python
hottest_annaba_days = np.where((T[:, 1] > T[:, 0]) & (T[:, 1] > T[:, 2]))[0]
# Résultat : [0, 1, 3, 4, 6] → Jours 1, 2, 4, 5, 7
print(hottest_annaba_days)
```
